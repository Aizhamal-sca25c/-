```python
import math

#Abykeeva_18_SCA-25C
x = 4.93
eps = 1e-6

# Приводим угол к промежутку [-pi/2, pi/2]
x = x - 2 * math.pi

term = x
sin_x = 0
k = 0
count = 0

while True:
    sin_x += term
    count += 1

    next_term = -term * x * x / ((2 * k + 2) * (2 * k + 3))

    if abs(next_term) < eps:
        break

    term = next_term
    k += 1

print("Приведенный угол:", x)
print("sin(x) =", sin_x)
print("Количество членов:", count)
print("Абсолютная погрешность:", abs(sin_x - math.sin(4.93)))
```

```python
#Abykeeva_19_SCA-25C
import math

x = 0.34
eps = 1e-8

t = 1
s = 0
k = 0
count = 0

while True:
    s += t
    count += 1

    next_term = -t * x * x / ((2 * k + 1) * (2 * k + 2))

    if abs(next_term) < eps:
        break

    t = next_term
    k += 1

print("Сумма =", s)
print("Количество членов =", count)
print("Абсолютная погрешность =", abs(s - math.cos(x)))
```

```python
#Abykeeva_20_SCA-25C
import math

t = 0.45
eps = 1e-8

# Вычисляем e^t
term_e = 1
sum_e = 0
k = 0
count_e = 0

while True:
    sum_e += term_e
    count_e += 1

    next_term = term_e * t / (k + 1)

    if abs(next_term) < eps:
        break

    term_e = next_term
    k += 1

# Вычисляем cos(t)
term_cos = 1
sum_cos = 0
k = 0
count_cos = 0

while True:
    sum_cos += term_cos
    count_cos += 1

    next_term = -term_cos * t * t / ((2 * k + 1) * (2 * k + 2))

    if abs(next_term) < eps:
        break

    term_cos = next_term
    k += 1

# F(t) = e^t * cos(t)
F = sum_e * sum_cos

# Проверка через библиотечные функции
F_library = math.exp(t) * math.cos(t)

print("F(t) =", F)
print("Число членов ряда e^t =", count_e)
print("Число членов ряда cos(t) =", count_cos)
print("Ошибка =", abs(F - F_library))
```
